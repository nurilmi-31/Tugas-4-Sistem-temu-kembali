```python
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import re

dokumen = [
    "Fakultas Teknik UNM akan melaksanakan kegiatan Seminar Nasional Teknologi Informasi pada hari Senin!",
    "Universitas Negeri Makassar mengumumkan bahwa pendaftaran Program Kreativitas Mahasiswa (PKM) Fakultas Teknik telah dibuka.",
    "Dalam rangka meningkatkan kompetensi mahasiswa, Fakultas Teknik UNM menyelenggarakan pelatihan Artificial Intelligence (AI) dan Machine Learning.",
    "Fakultas Teknik UNM mengadakan kegiatan kerja bakti dan pemeliharaan lingkungan kampus pada hari Jumat.",
    "Bagi mahasiswa Fakultas Teknik UNM yang akan mengikuti kegiatan Praktik Industri, diwajibkan untuk mengumpulkan dokumen administrasi sebelum tanggal 15 Oktober 2026."
]

stopword = StopWordRemoverFactory().create_stop_word_remover()
stemmer = StemmerFactory().create_stemmer()

for i, teks in enumerate(dokumen[:3], 1):

    print("\nDOKUMEN", i)

    # SEBELUM PREPROCESSING
    print("\nSEBELUM PREPROCESSING")
    print(teks)

    token_sebelum = re.findall(r'\w+', teks.lower())

    # Cleaning
    teks_clean = re.sub(r'[^a-zA-Z0-9\s]', '', teks)

    # Case Folding
    teks_lower = teks_clean.lower()

    # Tokenisasi
    tokens = re.findall(r'\w+', teks_lower)

    # Stopword Removal
    teks_no_stop = stopword.remove(teks_lower)

    # Stemming
    hasil_stem = stemmer.stem(teks_no_stop)

    token_sesudah = re.findall(r'\w+', hasil_stem)

    jumlah_sebelum = len(token_sebelum)
    jumlah_sesudah = len(token_sesudah)

    persentase_pengurangan_token = (
        (jumlah_sebelum - jumlah_sesudah) / jumlah_sebelum
    ) * 100

    # SESUDAH PREPROCESSING
    print("\nSESUDAH PREPROCESSING")
    print(hasil_stem)

    # HASIL SETIAP TAHAP
    print("\nHASIL SETIAP TAHAP")

    print("\n1. Cleaning:")
    print(teks_clean)

    print("\n2. Case Folding:")
    print(teks_lower)

    print("\n3. Tokenisasi:")
    print(tokens)

    print("\n4. Stopword Removal:")
    print(teks_no_stop)

    print("\n5. Stemming:")
    print(hasil_stem)

    # JUMLAH TOKEN
    print("\nPERHITUNGAN TOKEN")
    print("Jumlah token sebelum preprocessing :", jumlah_sebelum)
    print("Jumlah token setelah preprocessing :", jumlah_sesudah)
    print("Persentase pengurangan token       : {:.2f}%".format(persentase_pengurangan_token))


print("\nANALISIS")
print("Preprocessing meningkatkan kualitas data untuk sistem Information Retrieval (IR)"
      "\ndengan membersihkan teks dari tanda baca dan ketidakkonsistenan huruf, menghapus kata-kata umum (stopword)," 
      "\nserta mengubah kata berimbuhan menjadi bentuk dasarnya melalui stemming. "
      "\nProses ini membuat data menjadi lebih terstruktur, mengurangi jumlah token yang tidak penting," 
      "\ndan membantu sistem IR mencocokkan kata dalam dokumen dengan kata pada query secara lebih efektif," 
      "\nsehingga proses pencarian informasi menjadi lebih relevan")
```
